You are given a list of outcomes and their respective probabilities as two lists. Write a Python function to compute the expected value of the random variable.



Use:


Expected Value (E(X)) = Sum of [Each outcome (xi) × Its corresponding probability P(xi)].

Or in simple terms:

Expected Value = Σ (Outcome × Probability)


If the sum of probabilities is not 1, return "Invalid input".



Input Format:



A list containing two lists:



[ [x1, x2, ..., xn], [p1, p2, ..., pn] ]



The outcomes and probabilities lists have been extracted already in the starter code.

outcomes = [x1, x2, ..., xn]

probabilities = [p1, p2, ..., pn]



Output Format:

Expected Value: A float rounded to 2 decimal places or "Invalid input"





Sample Input:

[[1, 2, 3], [0.2, 0.5, 0.3]]



Sample Output:

2.1

```python
import ast

data = ast.literal_eval(input())
outcomes, probabilities = data[0], data[1]

def expected_value(outcomes, probabilities):
    if len(outcomes) != len(probabilities):
        return "Invalid input"

    if any(p < 0 or p > 1 for p in probabilities):
        return "Invalid input"

    if abs(sum(probabilities) - 1.0) > 1e-9:
        return "Invalid input"

    return round(sum(x * p for x, p in zip(outcomes, probabilities)), 2)

print(expected_value(outcomes, probabilities))
```



You are given a normally distributed variable with mean μ and standard deviation σ, and a value x.

Write a Python function to calculate the corresponding z-score, defined as:

z = (x - μ) / σ

Round the result to 2 decimal places. If σ ≤ 0, return "Invalid input".



Input Format:

A tuple: (x, mu (μ), sigma (σ) )



Output Format:

A float rounded to 2 decimal places or "Invalid input"





Sample Input:

(75, 70, 5)



Sample Output:

1.0

```python
import math, ast
data = ast.literal_eval(input())

# Input has been taken for you as a tuple

# Write your code below

x, mu, sigma = data

def calculate_z_score(x, mu, sigma):
    if sigma <= 0:
        return "Invalid input"

    z = (x - mu) / sigma
    return round(z, 2)

print(calculate_z_score(x, mu, sigma))
```

Given two vectors u and v in Rn space, write a Python function that:

Computes the magnitude of each vector
Computes the dot product of the two vectors


If the vectors are not of the same length, return "Invalid input".





Input Format:

A list containing two lists of floats: [list u, list v]

Note that the length of the strings is not fixed



Output Format:

A tuple with three floats:

(||u||, ||v||, u ⋅ v) magnitudes of both vectors and their dot product, rounded to 2 decimal places.




Or "Invalid input" if dot product is not possible between the two vectors





Sample Input:

[[1, 2, 3], [4, 5, 6]]



Sample Output:

(3.74, 8.77, 32.0)

```python
import math, ast
data = ast.literal_eval(input())
u, v = data

# Input has been taken for you and the lists u and v have been extracted
# Write your code below


def vector_operations(u, v):
    if len(u) != len(v):
        return "Invalid input"

    # Calculate magnitudes
    mag_u = math.sqrt(sum(x**2 for x in u))
    mag_v = math.sqrt(sum(x**2 for x in v))

    # Calculate dot product
    dot_product = sum(x * y for x, y in zip(u, v))

    return (round(mag_u, 2), round(mag_v, 2), round(dot_product, 2))

print(vector_operations(u, v))
```

Given the function:

f(x, y) = x²y + y³

Write a Python program to compute the partial derivatives (∂f/∂x, ∂f/∂y) at a given point (x, y).

Return both values rounded to 2 decimal places.

Input Format:

- Two real numbers as a tuple: (x, y).

Output Format:

- A tuple containing both partial derivatives, rounded to 2 decimal places.

Sample Input:
(1, 2)

Sample Output:
(4.0, 13.0)

```python
import math, ast
data = ast.literal_eval(input())
x, y = data

# The input tuple has been taken for you

# Write your code below

# Partial derivative with respect to x: 2xy
df_dx = 2 * x * y

# Partial derivative with respect to y: x² + 3y²
df_dy = x**2 + 3 * y**2

print((round(df_dx, 2), round(df_dy, 2)))
```
